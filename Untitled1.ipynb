```python
import cmath
import math


z = complex(4,4)
n = 3

if n <= 0:
  raise ValueError(" n must be a positive integer. ")
else:
  r, theta = cmath.polar(z)

  root_r = r**(1/n)
  roots = []

  print(f"--- The {n} roots of {z} ---")

  for k in range(n):
    root_theta = (theta + 2*math.pi*k)/n
    root_cartesian = cmath.rect(root_r, root_theta)

    roots.append(root_cartesian)

    print(f"\nRoot {k+1}:")
    print(
        f"  Cartesian form: {root_cartesian.real:.4f} + {root_cartesian.imag:.4f}j"
      )
    print(
        f"  Polarform: Magnitude = {root_r:.4f}, Angle = {root_theta:.4f} rad"
      )






```
